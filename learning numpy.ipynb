Level 4 -- proper mini project

```python
import numpy as np

student = np.array(["Rahul", "Aman","Neha","Ravi"])

marks = np.array([
    [78, 85, 67, 90, 72],
    [55, 65, 70, 60, 80],
    [92, 88, 95, 85, 90],
    [35, 45, 40, 38, 50]
])
print("student:")
print(student)

print("\nMarks:")
print(marks)

total = np.sum(marks, axis=1)
print("Total marks :", total)

average = np.mean(marks, axis=1)
print("Average:", average)

highest = np.max(marks, axis=1)
print("Highest:", highest)

lowest = np.min(marks, axis=1)
print("lowest:", lowest)

```
